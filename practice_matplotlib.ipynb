```python
import matplotlib.pyplot as plt
```

```python
# CREATE CHART
plt.figure(figsize=(4,3))

X=[1,2,3,4,5]
Y=[10,25,30,45,50]
plt.plot(X,Y)
plt.show()
```

```python
# Customise chart
plt.figure(figsize=(4,3))
plt.plot(X,Y,color='green',marker='o',linestyle='dashed',linewidth=2,markersize=12)
plt.title("SALES GRAPH")
plt.xlabel("MONTHS")
plt.ylabel("SALES")
plt.show()
```

```python
# ADVANCED LINE CHART -- MULTIPLE LINE & LEGENDS
```

```python
plt.figure(figsize=(4,3))
X=[1,2,3,4,5]
Y1=[10,25,30,45,50]
Y2=[15,20,35,40,55]
plt.plot(X,Y1,label='SALES 2024')
plt.plot(X,Y2,label='SALES 2025')
plt.title("YOY SALES")
plt.xlabel("Months")
plt.ylabel("SALES")
plt.legend()
plt.show()
```

```python
# BAR CHARTS
plt.figure(figsize=(4,3))
X=['A','B','C','D','E']
Y=[10,20,30,45,50]
plt.bar(X,Y)
plt.show()
```

```python
# HISTOGRAM CHART
plt.figure(figsize=(4,3))
import random
data=[ random.randint(1,20) for i in range(500)]
plt.hist(data, bins=10)
plt.title("HISTOGRAM EXAMPLE")
plt.show()
```

```python
# PIE CHART(used to show part to hole relatinship)
categories=['A','B','C','D','E']
sales=[10,20,30,45,50]
plt.pie(sales, labels=categories, autopct='%1.1f%%')
plt.title("PIE CHART EXAMPLE")
plt.show()
```

```python
# SCATTER PLOT
plt.figure(figsize=(4,3))
Y1=[10,25,30,45,50]
Y2=[15,20,35,40,55]
plt.scatter(Y1,Y2)
plt.title("SCATER PLOT EXAMPLE")
plt.show()
```

```python
# SUBPLOT
# data 1 bar chart
plt.figure(figsize=(6,3))
categories=['MON','TUE','WED','THU','FRI']
sales=[10,20,33,43,50]

# data 2 scatter plot
Y1=[10,20,30,40,50,]
Y2=[15,25,35,45,55]

# first plot bar chart

plt.subplot(2,1,1)# row,coloumn,position
plt.bar(categories,sales)
plt.xlabel("WEEK DAYS")
plt.ylabel("SALES")
plt.title("DAILY SALES")


# second plot scatter plot 
plt.subplot(2,1,2)
plt.scatter(Y1,Y2)
plt.title("User Example")
plt.xlabel("USER 1")
plt.ylabel("USER 2")
plt.show()
```

```python
# MATPLOTLIB WITH REAL DATA PYTHON
import matplotlib.pyplot as plt
import pandas as pd
data={
    'month':['JAN','FEB','MAR','APR','MAY'],
    'sales':[12000 ,25000 ,35000 ,45000, 55000]
}
df=pd.DataFrame(data)
df

```

```python
plt.bar(df['month'],df['sales'])
plt.title("matplotlib with pandas")
plt.xlabel("month")
plt.ylable("sales")

plt.show()
```
